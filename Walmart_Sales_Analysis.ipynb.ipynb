```python
df['unit_price'] = df['unit_price'].str.replace('$', '', regex=False)
```

```python
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')
```

```python
df['unit_price'] = df['unit_price'].fillna(df['unit_price'].median())
```

```python
df['quantity'] = df['quantity'].fillna(df['quantity'].median())
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df['total_sales'] = df['unit_price'] * df['quantity']
df['total_sales']
```

```python
df['date'] = pd.to_datetime(df['date'], dayfirst=True)
```

```python
#Category-wise Revenue Analysis
category_sales = df.groupby('category')['total_sales'].sum().sort_values(ascending=False)

category_sales
```

```python
# Overall Business Performance (KPI Summary)

total_revenue = df['total_sales'].sum()
total_transactions = len(df)
average_order_value = df['total_sales'].mean()
average_rating = df['rating'].mean()

print("===== WALMART SALES OVERVIEW =====")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Transactions: {total_transactions:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Average Customer Rating: {average_rating:.2f}")
```

```python
#Top Categories by Revenue

category_sales = (
    df.groupby('category')['total_sales']
    .sum()
    .sort_values(ascending=False)
)
category_sales
```

```python
category_sales.plot(
    kind='bar',
    figsize=(8, 6)
)
plt.title('Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue')

plt.xticks(rotation=45)
plt.show()
```

```python
#Best & Worst Performing 

worst_category = category_sales.idxmin()

print(f"🏆 Best Performing Category: {best_category}")
print(f"📉 Worst Performing Category: {worst_category}")
```

```python
#Branch Performance Analysis
branch_performance = df.groupby('Branch').agg(
    Total_Revenue=('total_sales', 'sum'),
    Total_Orders=('invoice_id', 'count'),
    Average_Rating=('rating', 'mean'),
    Average_Quantity=('quantity', 'mean')
).sort_values(
    by='Total_Revenue',
    ascending=False
)

branch_performance
```

```python
#Top Cities by Revenue
city_performance = (
    df.groupby('City')
    .agg(
        Total_Revenue=('total_sales', 'sum'),
        Total_Orders=('invoice_id', 'count'),
        Average_Rating=('rating', 'mean')
    )
    .sort_values(
        by='Total_Revenue',
        ascending=False
    )
)

city_performance.head(10)
```

```python
city_performance.head(10)['Total_Revenue'].plot(
    kind='bar',
    figsize=(8, 6)
)

plt.title('Top 10 Cities by Revenue')
plt.xlabel('City')
plt.ylabel('Total Revenue')

plt.xticks(rotation=45)
plt.show()
```

```python
#Revenue vs Profit Analysis
df['profit_margin'].describe()
```

```python
df['profit'] = (
    df['total_sales'] * df['profit_margin']
)
```

```python
category_profit = (
    df.groupby('category')['profit']
    .sum()
    .sort_values(ascending=False)
)

category_profit
```

```python
category_profit.plot(
    kind='bar',
    figsize=(8, 3)
)

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')

plt.xticks(rotation=45)
plt.show()
```

```python
# High-Value Transactions Analysis
threshold = df['total_sales'].quantile(0.90)

high_value_transactions = df[
    df['total_sales'] > threshold
]

print("High-Value Transaction Threshold:", threshold)
print("Number of High-Value Transactions:", len(high_value_transactions))
```

```python
#Category-wise Analysis:
high_value_category = (
    high_value_transactions
    .groupby('category')['total_sales']
    .sum()
    .sort_values(ascending=False)
)

high_value_category
```

```python
high_value_category.plot(
    kind='bar',
    figsize=(8, 3)
)

plt.title('High-Value Transaction Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue')

plt.xticks(rotation=45)
plt.show()
```

```python
#Rating vs Sales Analysis
rating_sales_analysis = df.groupby('category').agg(
    Average_Rating=('rating', 'mean'),
    Total_Revenue=('total_sales', 'sum')
)

rating_sales_analysis
```

```python
plt.figure(figsize=(6, 4))

plt.scatter(
    rating_sales_analysis['Average_Rating'],
    rating_sales_analysis['Total_Revenue']
)

plt.title('Average Rating vs Total Revenue')
plt.xlabel('Average Customer Rating')
plt.ylabel('Total Revenue')

plt.show()
```

```python
# Payment Method Analysis
payment_analysis = df.groupby('payment_method').agg(
    Total_Revenue=('total_sales', 'sum'),
    Total_Transactions=('invoice_id', 'count'),
    Average_Order_Value=('total_sales', 'mean')
).sort_values(
    by='Total_Revenue',
    ascending=False
)

payment_analysis
```

```python
payment_analysis['Total_Revenue'].plot(
    kind='bar',
    figsize=(8, 3)
)

plt.title('Revenue by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Revenue')

plt.xticks(rotation=0)
plt.show()
```
